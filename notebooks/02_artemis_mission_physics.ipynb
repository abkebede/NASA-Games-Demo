## Day 1 - Launch and orbital motion

For a circular orbit,

\[
v_{\rm orb}=\sqrt{\frac{\mu}{r}}.
\]

Students calculate orbital speed and examine how it changes with orbital radius.


```python
import numpy as np
import matplotlib.pyplot as plt

mu_earth = 398600.44 # km^3/s^2
r = np.linspace(6500, 16000, 200) # km
v = np.sqrt(mu_earth/r)

print("Orbital speed at r = 6571 km:", np.sqrt(mu_earth/6571), "km/s")

plt.figure(figsize=(7,4))
plt.plot(r,v)
plt.xlabel("Orbital radius (km)")
plt.ylabel("Circular orbital speed (km/s)")
plt.title("Orbital speed versus orbital radius")
plt.grid(True,alpha=0.3)
plt.show()
```

## Day 3 - Deep-space communications

Signal propagation is limited by the speed of light:

\[
t=\frac{R}{c}.
\]

At lunar distance, the delay is large enough that communication cannot be treated as instantaneous.


```python
c = 299792.458 # km/s
R = 384400.0 # km
t = R/c
print("Earth-Moon one-way light time:", t, "s")

ranges = np.linspace(1e4,4e5,200)
times = ranges/c

plt.figure(figsize=(7,4))
plt.plot(ranges,times)
plt.xlabel("Range (km)")
plt.ylabel("One-way light time (s)")
plt.title("Communication delay versus range")
plt.grid(True,alpha=0.3)
plt.show()
```

## Day 8 - Thermal control

A simplified radiator model is

\[
P_{\rm rad}=\epsilon\sigma A(T^4-T_{\rm sink}^4).
\]

This activity shows why temperature changes strongly affect radiative heat rejection.


```python
sigma = 5.670374419e-8
epsilon = 0.85
A = 35.0 # m^2
T_sink = 100.0 # K
T = np.linspace(240,360,200)
P = epsilon*sigma*A*(T**4-T_sink**4)/1000

plt.figure(figsize=(7,4))
plt.plot(T,P)
plt.xlabel("Radiator temperature (K)")
plt.ylabel("Radiative heat rejection (kW)")
plt.title("Radiative heat rejection versus temperature")
plt.grid(True,alpha=0.3)
plt.show()
```
